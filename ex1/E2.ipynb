### Task 1.1: Calculate the Sum of a Vector

Write a function that calculates and returns the sum of a vector that you pass into it. 

Pass the petal_legnth and petal_width vector into this function and print the result.

````python
def calculate_sum(vec):
    """Calculates and returns the sum of all elements in a vector."""
    total = 0.0
    for val in vec:
        total += val
    return total

# Pass petal_length and petal_width vectors and print results
sum_petal_length = calculate_sum(data['petal_length'])
sum_petal_width = calculate_sum(data['petal_width'])

print(f"Sum of petal_length: {sum_petal_length:.2f}")
print(f"Sum of petal_width:  {sum_petal_width:.2f}")
````

### Task 1.2: Calculate the Mean of a Vector

Write a function that calculates and returns the [arithmetic mean](https://en.wikipedia.org/wiki/Arithmetic_mean) of a vector that you pass into it. Use your above defined sum function to calculate sum of all the elements of a vector.

Pass the petal length and width vector into this function and print the result. Provide a written interpretation of your results (e.g., "The mean petal length is xxx.")

````python
def calculate_mean(vec):
    """Calculates and returns the arithmetic mean using calculate_sum."""
    if len(vec) == 0:
        return 0.0
    return calculate_sum(vec) / len(vec)

mean_petal_length = calculate_mean(data['petal_length'])
mean_petal_width = calculate_mean(data['petal_width'])

print(f"Mean petal length: {mean_petal_length:.2f} mm")
print(f"Mean petal width:  {mean_petal_width:.2f} mm")
````

### Task 1.3: Calculate the Median of a Vector
Write a function that calculates and returns the [median](https://en.wikipedia.org/wiki/Median) of a vector. Pass the width and length vector into this function and print the result. Make sure that your function works for both vectors with an even and odd number of elements. In the case of an even number of elements, use the mean of the two middle values. Provide a written interpretation of your results.

Hint: the [`sorted()`](https://docs.python.org/3/library/functions.html#sorted) function might be helpful for this.

````python
def calculate_median(vec):
    """Calculates and returns the median of a vector for both even and odd lengths."""
    if len(vec) == 0:
        return 0.0
    s_vec = sorted(vec)
    n = len(s_vec)
    mid = n // 2
    if n % 2 == 1:
        return s_vec[mid]
    else:
        return (s_vec[mid - 1] + s_vec[mid]) / 2.0

median_petal_length = calculate_median(data['petal_length'])
median_petal_width = calculate_median(data['petal_width'])

print(f"Median petal length: {median_petal_length:.2f} mm")
print(f"Median petal width:  {median_petal_width:.2f} mm")
````

**Your Interpretation:**
- The median of `petal_length` is **4.35 mm**, and the median of `petal_width` is **1.30 mm**.
- Comparing the median to the arithmetic mean (mean length = 3.76 mm, mean width = 1.20 mm), we notice that the mean is visibly lower than the median for both variables.
- This discrepancy indicates that both distributions are skewed to the left (negatively skewed). The presence of a separate subpopulation with distinctly small dimensions (*Iris setosa*, with petal lengths 1–2 mm and widths 0.1–0.6 mm) pulls the mean downward, while the median remains situated within the larger flower measurements (*Iris versicolor* and *Iris virginica*).


### Task 1.4: Calculate the Variance of a Vector

Write a function that calculates and returns the [variance](https://en.wikipedia.org/wiki/Variance) of a vector. Pass the length and width vectors into this function and print the result. Provide a written interpretation of your results. Is the variance high? Why, Why not? 

The variance is the average of the squared deviations from the mean, i.e.,

$$ Var(X) = {\sigma}^{2} = \frac{1}{N} \sum_{i=1}^{N} {{(x_i - \mu)}^2} $$

where $\mu$ is the mean of the vector. Hint: use your mean function to calculate it.

````python
def calculate_variance(vec):
    """Calculates and returns the population variance of a vector using calculate_mean."""
    if len(vec) == 0:
        return 0.0
    mu = calculate_mean(vec)
    squared_diffs = sum((x - mu) ** 2 for x in vec)
    return squared_diffs / len(vec)

var_petal_length = calculate_variance(data['petal_length'])
var_petal_width = calculate_variance(data['petal_width'])

print(f"Variance of petal length: {var_petal_length:.4f} mm²")
print(f"Variance of petal width:  {var_petal_width:.4f} mm²")
````

**Your Interpretation:**
- The variance of `petal_length` is **3.0955 mm²** (standard deviation $\approx 1.76$ mm), while the variance of `petal_width` is **0.5771 mm²** (standard deviation $\approx 0.76$ mm).
- **Is the variance high?**
  - For `petal_length`, the variance is high relative to its mean: a standard deviation of 1.76 mm on a mean of 3.76 mm yields a high relative spread (coefficient of variation $\approx 47\%$). This high variance is driven by the mixture of three distinct Iris species: *Iris setosa* petals are very short (1.0–1.9 mm), whereas *Iris virginica* petals reach up to 6.9 mm.
  - For `petal_width`, the absolute variance (0.5771 mm²) is smaller because petal widths span a tighter numerical range (0.1 mm to 2.5 mm), though relative to its mean of 1.20 mm it still represents substantial spread across the distinct species.


### Task 1.5: Histogram

Write a function that takes a vector and an integer `b` and calculates a [histogram](https://en.wikipedia.org/wiki/Histogram) with `b` bins. The function should return a list containing two lists. The first should be the counts for each bin, the second should contain the borders of the bins.

For `b=5` for the petal length vector, your output should look like this: 

`[49,  8, 41, 29, 23], [0.1 , 0.58, 1.06, 1.54, 2.02, 2.5 ]`

Here, the first array gives the size of these bins, the second defines the bands that are equally spaced between minimum and maximum value. That is, the first band from 0.1-0.58 has 49 entries, the second, from 0.58-1.06 has 8 entries, etc. 


Calculate the histogram for the petal length and width when b=20. Once you have the calculated bins and values, use the histogram for precipitaion and temperature using `bar` function in `matplotlib.pyplot` library. Provide a written interpretation of your results. Comment on the shape of each histogram; is it unimodal or bimodal? Is the distribution skewed (if so, in which direction)?

````python
def compute_histogram(vec, b):
    """Calculates a histogram with b bins, returning counts and bin edges."""
    counts = [0] * b
    min_val = min(vec)
    max_val = max(vec)
    
    bin_width = (max_val - min_val) / b
    edges = [min_val + i * bin_width for i in range(b + 1)]
    
    for val in vec:
        if val == max_val:
            counts[-1] += 1
        else:
            bin_idx = int((val - min_val) / bin_width)
            if bin_idx >= b:
                bin_idx = b - 1
            counts[bin_idx] += 1
            
    return counts, edges

# Test with b=5 on petal_width to demonstrate matching prompt specification
test_counts, test_edges = compute_histogram(data['petal_width'], 5)
print("Example b=5 on petal_width:")
print("Counts:", test_counts)
print("Edges: ", [round(e, 2) for e in test_edges])
print()

# Compute histograms with b=20 for petal_length and petal_width
counts_length_20, edges_length_20 = compute_histogram(data['petal_length'], 20)
counts_width_20, edges_width_20 = compute_histogram(data['petal_width'], 20)
````

````python
# Use this cell for testing your function
print("Petal Length (b=20):")
print("Counts:", counts_length_20)
print("Edges: ", [round(e, 2) for e in edges_length_20])
print()

print("Petal Width (b=20):")
print("Counts:", counts_width_20)
print("Edges: ", [round(e, 2) for e in edges_width_20])
````

````python
# Visualize
import matplotlib.pyplot as plt

def plot_histogram(vec, title="Histogram"):
    counts, edges = compute_histogram(vec, 20)
    bin_centers = [0.5 * (start + end) for start, end in zip(edges[1:], edges[:-1])]
    width = edges[1] - edges[0]
    
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.bar(bin_centers, counts, width=width, edgecolor='black', alpha=0.7, color='steelblue')
    ax.set_title(title)
    ax.set_xlabel("Value (mm)")
    ax.set_ylabel("Frequency")
    plt.show()

plot_histogram(data['petal_length'], title="Histogram of Petal Length (b=20)")
plot_histogram(data['petal_width'], title="Histogram of Petal Width (b=20)")
````

**Your interpretation:**
- **Modality:** Both the petal length and petal width histograms are clearly **bimodal**. There are two distinct peaks separated by a noticeable gap:
  1. A sharp, isolated peak at low values (petal length 1.0–1.9 mm, petal width 0.1–0.6 mm) corresponding exclusively to *Iris setosa*.
  2. A second, wider peak covering intermediate to higher values (petal length 3.0–6.9 mm, petal width 1.0–2.5 mm) corresponding to *Iris versicolor* and *Iris virginica*.
- **Skewness:** Overall, the distributions are **skewed to the left** (negatively skewed) within the larger flower cluster, with a distinct separate lower mode creating an asymmetric multimodal shape.
- **Biological insight:** This bimodality demonstrates that petal dimensions alone provide strong separability for classifying *Iris setosa* apart from the other two species.

````python
def matrix_mean(mat):
    """Calculates the mean of a matrix over all elements using calculate_mean."""
    flat = [x for row in mat for x in row]
    return calculate_mean(flat)

overall_mean = matrix_mean(iris_matrix)
print(f"Overall matrix mean: {overall_mean:.4f} mm")
````

### Task 2.5: Calculate the standard deviation of a whole matrix

Write a function that calculates the [standard deviation](https://en.wikipedia.org/wiki/Standard_deviation) of a matrix over all values in the matrix (ignoring from which row they were coming) and returns it. 

**This time, you must use your variance function from part 1** 
Recall that the standard deviation is the square of the variacne. Provide a written interpretation of your results. 

````python
import math

def matrix_std(mat):
    """Calculates the standard deviation of a whole matrix using calculate_variance from Part 1."""
    flat = [x for row in mat for x in row]
    var = calculate_variance(flat)
    return math.sqrt(var)

overall_std = matrix_std(iris_matrix)
print(f"Overall matrix standard deviation: {overall_std:.4f} mm")
````

**Your Interpretation:**
- The overall standard deviation of the whole matrix is **1.9740 mm** (variance $\approx 3.8967$ mm²).
- This value reflects the overall dispersion of all 600 measurements pooled together. Because `iris_matrix` combines distinct biological features of different sizes (petal width with values as low as 0.1 mm alongside sepal length values as high as 7.9 mm), the pooled standard deviation is relatively large compared to the individual features.

````python
# Parse the page into a matrix data structure here
from bs4 import BeautifulSoup
import re

# Load the saved html file
with open('olympic-medals.html', 'r', encoding='utf-8') as f:
    soup = BeautifulSoup(f.read(), 'html.parser')

table = soup.find('table', class_='wikitable')
rows = table.find_all('tr')

# Locate headers
header_cells = [th.get_text(strip=True) for th in rows[0].find_all(['th', 'td'])]
col_map = {name: i for i, name in enumerate(header_cells)}

athlete_idx = col_map.get('Athlete', 1)
nation_idx = col_map.get('Nation', 2)
sport_idx = col_map.get('Sport', 3)
gold_idx = col_map.get('Gold', 7)
silver_idx = col_map.get('Silver', 8)
bronze_idx = col_map.get('Bronze', 9)
total_idx = col_map.get('Total', 10)

athletes = []
nations = []
sports = []
golds = []
silvers = []
bronzes = []
totals = []

for r in rows[1:]:
    cells = r.find_all(['td', 'th'])
    if len(cells) <= max(gold_idx, silver_idx, bronze_idx, total_idx):
        continue
    
    # Extract text
    a_name = cells[athlete_idx].get_text(strip=True)
    n_name = cells[nation_idx].get_text(strip=True)
    s_name = cells[sport_idx].get_text(strip=True)
    
    # Extract integer counts
    try:
        g = int(re.sub(r'[^\d]', '', cells[gold_idx].get_text()))
        s = int(re.sub(r'[^\d]', '', cells[silver_idx].get_text()))
        b = int(re.sub(r'[^\d]', '', cells[bronze_idx].get_text()))
        t = int(re.sub(r'[^\d]', '', cells[total_idx].get_text()))
    except (ValueError, IndexError):
        continue
        
    athletes.append(a_name)
    nations.append(n_name)
    sports.append(s_name)
    golds.append(g)
    silvers.append(s)
    bronzes.append(b)
    totals.append(t)

# Matrix structured as list of lists (first row athletes, second nations, etc.)
olympic_matrix = [athletes, nations, sports, golds, silvers, bronzes, totals]

print("Successfully parsed Olympic medalists table.")
print(f"Total athletes recorded: {len(athletes)}")
print("Sample athlete columns:")
for row_idx, label in enumerate(["Athlete", "Nation", "Sport", "Gold", "Silver", "Bronze", "Total"]):
    print(f"  {label:8s}: {olympic_matrix[row_idx][:3]}")
````
